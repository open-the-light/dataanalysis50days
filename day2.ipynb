```python
def reverse_array(array):
    i = 0
    j = len(array) - 1
    while i < j:
        t = array[i]
        array[i] = array[j]
        array[j] = t
        i += 1
        j -= 1
    return array
```

```python
print(reverse_array(arr1))
```
