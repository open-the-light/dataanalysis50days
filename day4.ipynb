```python
import numpy as np
```

```python
rng = np.random.default_rng(seed=10)
```

```python
r1 = rng.integers(10, 20, size=(2, 3))
r2 = rng.integers(10, 20, size=(1, 3))
print(r1)
print(r2)
```

```python
r1 + r2
```

```python
np.dot(r1, r2.T)
```

```python
r3 = rng.integers(10, size=(100,))
```

```python
import matplotlib.pyplot as plt
plt.hist(r3, bins=10)
plt.show()
```

```python
print(np.median(r3))
v, c = np.unique(r3, return_counts=True)
print(v[np.argmax(c)])
```

```python
r4 = (rng.random(size=100), rng.random(size=100), rng.random(size=100))
```

```python
fig = plt.figure()
ax = fig.add_subplot(projection='3d')
ax.scatter(*r4)
plt.show()
```
