```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
rng = np.random.default_rng(seed = 10)
```

```python
r1 = rng.random((3, 4))
r1
```

```python
r2 = rng.integers(10, size=(3, 4))
r2
```

```python
r3 = rng.standard_normal(1000)
r3[:5]
```

```python
plt.hist(r3, bins=25)
plt.show()
```

```python
arr = ["Orange", "Apple", "Pear"]
r4 = rng.choice(arr, size=(3, 4))
r4
```
